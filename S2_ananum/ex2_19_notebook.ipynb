```python
import numpy as np
import matplotlib.pyplot as plt
```

### Paramètres du modèle ###

```python
lambda_val = 1j
T = 100 * 2 * np.pi
h = 2 * np.pi / 100
N = int(T / h)
x0 = 1.0
t = np.linspace(0, T, N+1)
```

### Définition des schémas numériques ###

```python
# Fonction pour le système différentiel
def system(x, t, lambda_val):
    return lambda_val * x

# Fonction pour le schéma d'Euler explicite
def euler_explicite(x0, t, h, lambda_val):
    n = len(t)
    x = np.zeros(n, dtype = complex)
    x[0] = x0

    for i in range(1, n):
        x[i] = x[i-1] + h * system(x[i-1], t[i-1], lambda_val)

    return x


def euler_implicite(x0, t, h, lambda_val):
    n=len(t)
    x = np.zeros(n, dtype = complex)
    x[0]=x0
    for i in range(1,n):
        x[i] = x[i-1]/(1-lambda_val*h)
    return x
```

### Approximation ###

```python
# Résolution du système avec Euler explicite
sol_ee = euler_explicite(x0, t, h, lambda_val)

sol_ei = euler_implicite(x0, t, h, lambda_val)

# Solution exacte
solution_exacte = x0 * np.exp(lambda_val * t)
```

### Graphique schéma EE ###

```python
# Tracé de la partie réelle
plt.figure(figsize=(12, 6))
plt.plot(t, sol_ee.real, label='Euler Explicite (Réel)', linestyle='--')
plt.plot(t, solution_exacte.real, label='Solution Exacte (Réel)', color='red')
plt.xlabel(r'$t$')
plt.ylabel(r'Partie Réelle de $x(t)$')
plt.xlim(0,60)
plt.ylim(-2,2)
plt.legend()
plt.title('Stabilité de la méthode d\'Euler explicite (Partie Réelle)')
plt.grid(True)
plt.show()

# Tracé de la partie imaginaire
plt.figure(figsize=(12, 6))
plt.plot(t, sol_ee.imag, label='Euler Explicite (Imaginaire)', linestyle='--')
plt.plot(t, solution_exacte.imag, label='Solution Exacte (Imaginaire)', color='red')
plt.xlabel('Temps')
plt.ylabel(r'Partie Imaginaire de $x(t)$')
plt.legend()
plt.title('Stabilité de la méthode d\'Euler explicite (Partie Imaginaire)')
plt.grid(True)
plt.show()
```

### Graphique schéma EI ###

```python
# Tracé de la partie réelle
plt.figure(figsize=(12, 6))
plt.plot(t, sol_ei.real, label='Euler Explicite (Réel)', linestyle='--')
plt.plot(t, solution_exacte.real, label='Solution Exacte (Réel)', color='red')
plt.xlabel(r'$t$')
plt.ylabel(r'Partie Réelle de $x(t)$')
plt.ylim(-2,2)
plt.legend()
plt.title('Stabilité de la méthode d\'Euler implicite (Partie Réelle)')
plt.grid(True)
plt.show()

# Tracé de la partie imaginaire
plt.figure(figsize=(12, 6))
plt.plot(t, sol_ei.imag, label='Euler Implicite (Imaginaire)', linestyle='--')
plt.plot(t, solution_exacte.imag, label='Solution Exacte (Imaginaire)', color='red')
plt.xlabel(r'$t$')
plt.ylabel(r'Partie Imaginaire de $x(t)$')
plt.ylim(-2,2)
plt.legend()
plt.title('Stabilité de la méthode d\'Euler implicite (Partie Imaginaire)')
plt.grid(True)
plt.show()
```
