**Interprétation des coefficients d'une régression linéaire.** (en lien avec Frisch-Waugh). 

```python
from sklearn.datasets import fetch_openml
import pandas as pd
survey = fetch_openml(data_id=534, as_frame=True, parser='auto')
```

```python
df = survey.data[survey.feature_names]
df.describe(include="all")
```

```python
df.info()
```

```python
categorical_features=df.select_dtypes(include='category')
categorical_features
```

```python
from sklearn.preprocessing import OneHotEncoder
encoder=OneHotEncoder(drop='first',sparse_output=False)
array_cat_binary=encoder.fit_transform(categorical_features)
array_cat_binary
```

```python
encoded_col_names = encoder.get_feature_names_out(categorical_features.columns)
encoded_col_names
```

```python
# Création d'un DataFrame à partir de l'array encodé, en utilisant l'index du DataFrame original
df_cat_binary = pd.DataFrame(array_cat_binary, columns=encoded_col_names, index=df.index)
```

```python
# dataframe final (avec variables numériques + variables catégorielles encodées)
df_final=pd.concat([df.select_dtypes(include='number'),df_cat_binary], axis=1)
df_final.describe()
```

```python
df_final.info()
```

```python
Y = survey.target
Y.describe()
```
